```python

import random

def dls(state, goal, limit, path, visited):
    if state == goal:
        return path

    if limit == 0:
        return None

    zero = state.index(0)
    row, col = zero // 3, zero % 3

    moves = [(-1, 0), (1, 0), (0, -1), (0, 1)]

    for dr, dc in moves:
        r, c = row + dr, col + dc

        if 0 <= r < 3 and 0 <= c < 3:
            pos = r * 3 + c
            board = list(state)
            board[zero], board[pos] = board[pos], board[zero]
            new_state = tuple(board)

            if new_state not in visited:
                visited.add(new_state)
                result = dls(new_state, goal, limit - 1,
                             path + [new_state], visited)

                if result:
                    return result

    return None


def ids(start, goal):
    depth = 0

    while True:
        visited = {start}
        result = dls(start, goal, depth, [start], visited)

        if result:
            return result

        depth += 1


def print_board(state):
    for i in range(0, 9, 3):
        print(*state[i:i+3])
    print()


goal = (1, 2, 3, 4, 5, 6, 7, 8, 0)

# Generate a random solvable initial state
start = list(goal)
zero = 8

for _ in range(10):
    row, col = zero // 3, zero % 3
    moves = []

    for dr, dc in [(-1, 0), (1, 0), (0, -1), (0, 1)]:
        r, c = row + dr, col + dc
        if 0 <= r < 3 and 0 <= c < 3:
            moves.append(r * 3 + c)

    pos = random.choice(moves)
    start[zero], start[pos] = start[pos], start[zero]
    zero = pos

start = tuple(start)

print("Initial State:")
print_board(start)

path = ids(start, goal)

print("Solution found! Moves:", len(path) - 1)

for i, state in enumerate(path):
    print("Step", i)
    print_board(state)
```
