```python
import random

def dfs(start, goal):
    stack = [(start, [start])]
    visited = set()

    while stack:
        state, path = stack.pop()

        if state == goal:
            return path

        if state in visited:
            continue

        visited.add(state)

        zero = state.index(0)
        row, col = zero // 3, zero % 3

        moves = [(-1, 0), (1, 0), (0, -1), (0, 1)]

        for dr, dc in moves:
            r, c = row + dr, col + dc

            if 0 <= r < 3 and 0 <= c < 3:
                pos = r * 3 + c
                board = list(state)
                board[zero], board[pos] = board[pos], board[zero]
                new_state = tuple(board)

                if new_state not in visited:
                    stack.append((new_state, path + [new_state]))

    return None


def print_board(state):
    for i in range(0, 9, 3):
        print(*state[i:i+3])
    print()


goal = (1, 2, 3, 4, 5, 6, 7, 8, 0)

# Generate a random solvable initial state
start = list(goal)
zero = 8

for _ in range(15):
    row, col = zero // 3, zero % 3
    moves = []

    for dr, dc in [(-1, 0), (1, 0), (0, -1), (0, 1)]:
        r, c = row + dr, col + dc
        if 0 <= r < 3 and 0 <= c < 3:
            moves.append(r * 3 + c)

    pos = random.choice(moves)
    start[zero], start[pos] = start[pos], start[zero]
    zero = pos

start = tuple(start)

print("Initial State:")
print_board(start)

path = dfs(start, goal)

if path:
    print("Solution found! Moves:", len(path) - 1)
    for i, state in enumerate(path):
        print("Step", i)
        print_board(state)
else:
    print("Solution not found.")
```
