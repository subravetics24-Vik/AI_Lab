```python
import random

def calculate_cost(state):
    n = len(state)
    attacks = 0
    for i in range(n):
        for j in range(i + 1, n):
            # Same row attack
            if state[i] == state[j]:
                attacks += 1
            # Diagonal attack
            elif abs(state[i] - state[j]) == abs(i - j):
                attacks += 1
    return attacks

def get_neighbors(state):
    neighbors = []
    n = len(state)
    for col in range(n):
        for row in range(n):
            if state[col] != row:
                neighbor = list(state)
                neighbor[col] = row
                neighbors.append(tuple(neighbor))
    return neighbors

def hill_climbing_n_queens(initial_state, max_iterations=1000):
    current_state = tuple(initial_state)
    current_cost = calculate_cost(current_state)

    print(f"Initial State: {current_state} | Initial Cost: {current_cost}\n")

    for iteration in range(1, max_iterations + 1):
        if current_cost == 0:
            print(f"Goal state reached in {iteration-1} steps!")
            return current_state, True

        neighbors = get_neighbors(current_state)
        if not neighbors:
            break

        neighbor_costs = [(neighbor, calculate_cost(neighbor)) for neighbor in neighbors]
        best_neighbor, best_cost = min(neighbor_costs, key=lambda x: x[1])

        print(f"Step {iteration}:")
        print(f"  Current State: {current_state} (Cost: {current_cost})")
        print(f"  Best Neighbor: {best_neighbor} (Cost: {best_cost})")

        if best_cost >= current_cost:
            print("\nStuck in local optimum or flat local maximum.")
            return current_state, False

        current_state = best_neighbor
        current_cost = best_cost

    return current_state, False
```

```python
initial_board = [0, 1, 2, 3]

final_state, success = hill_climbing_n_queens(initial_board)

print("\nFinal Board State:", final_state)
print("Success:", success)
```

```python
def display_board(state):
    n = len(state)
    for row in range(n):
        line = ""
        for col in range(n):
            if state[col] == row:
                line += " Q "
            else:
                line += " . "
        print(line)
    print()

print("Initial Board [0, 1, 2, 3]:")
display_board([0, 1, 2, 3])

print("Stuck/Final Board (1, 0, 2, 3):")
display_board((1, 0, 2, 3))
```
