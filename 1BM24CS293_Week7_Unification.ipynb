```python

def occurs_check(var, expr, subst):
    expr = substitute(expr, subst)

    if var == expr:
        return True

    if isinstance(expr, tuple):
        return any(occurs_check(var, arg, subst)
                   for arg in expr[1:])

    return False


def substitute(expr, subst):
    if isinstance(expr, str):
        while expr in subst:
            expr = subst[expr]
        return expr

    if isinstance(expr, tuple):
        return (expr[0],) + tuple(
            substitute(arg, subst) for arg in expr[1:]
        )

    return expr


def unify(x, y, subst):
    x = substitute(x, subst)
    y = substitute(y, subst)

    if x == y:
        return subst

    if isinstance(x, str) and x in variables:
        if occurs_check(x, y, subst):
            return None
        subst[x] = y
        return subst

    if isinstance(y, str) and y in variables:
        if occurs_check(y, x, subst):
            return None
        subst[y] = x
        return subst

    if isinstance(x, tuple) and isinstance(y, tuple):

        if x[0] != y[0]:
            return None

        if len(x) != len(y):
            return None

        for a, b in zip(x[1:], y[1:]):
            subst = unify(a, b, subst)

            if subst is None:
                return None

        return subst

    return None


# Convert user input into expressions
def parse_expression(s):
    name, args = s.split("(", 1)
    args = args.rstrip(")")
    arguments = tuple(a.strip() for a in args.split(","))

    return (name.strip(),) + arguments


# Take input from user
expr1 = input("Enter first expression: ")
expr2 = input("Enter second expression: ")

expr1 = parse_expression(expr1)
expr2 = parse_expression(expr2)

# Define variables
variables = {'x', 'y', 'z'}

# Perform unification
subst = {}
result = unify(expr1, expr2, subst)

if result is None:
    print("\nUnification Failed")
else:
    print("\nUnification Successful")
    print("Substitution Set:", result)

    print("Expression 1:", substitute(expr1, result))
    print("Expression 2:", substitute(expr2, result))
```
