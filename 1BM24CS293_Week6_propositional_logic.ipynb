```python

# Truth Table Enumeration Algorithm

# Function to check implication
def implies(a, b):
    return (not a) or b


# Function to check whether KB is true
def KB(P, Q, R):
    sentence1 = implies(Q, P)       # Q -> P
    sentence2 = implies(P, not Q)   # P -> ~Q
    sentence3 = Q or R              # Q V R

    return sentence1 and sentence2 and sentence3


# Function to evaluate the query
def evaluate_query(P, Q, R, query):
    if query == "R":
        return R

    elif query == "R -> P":
        return implies(R, P)

    elif query == "Q -> R":
        return implies(Q, R)


# Truth Table Enumeration
def truth_table_entails(query):

    print("\nP Q R | KB | Query")
    print("-------------------")

    entails = True

    # Generate all 8 truth assignments
    for P in [True, False]:
        for Q in [True, False]:
            for R in [True, False]:

                kb_value = KB(P, Q, R)
                query_value = evaluate_query(P, Q, R, query)

                print(int(P), int(Q), int(R), "|",
                      int(kb_value), "|", int(query_value))

                # Check only models where KB is true
                if kb_value and not query_value:
                    entails = False

    return entails


# Test all three queries
queries = ["R", "R -> P", "Q -> R"]

for query in queries:
    result = truth_table_entails(query)

    print("\nQuery:", query)

    if result:
        print("KB entails", query)
    else:
        print("KB does not entail", query)

    print("---------------------------")
```
