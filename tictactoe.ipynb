```python
board = [
    [' ', ' ', ' '],
    [' ', ' ', ' '],
    [' ', ' ', ' ']
]
```

```python
def check_win(player):

    for i in range(3):
        if board[i][0] == player and board[i][1] == player and board[i][2] == player:
            return True

    for i in range(3):
        if board[0][i] == player and board[1][i] == player and board[2][i] == player:
            return True

    if board[0][0] == player and board[1][1] == player and board[2][2] == player:
        return True

    if board[0][2] == player and board[1][1] == player and board[2][0] == player:
        return True

    return False


```

```python
def check_full():

    for i in range(3):
        for j in range(3):
            if board[i][j] == ' ':
                return False
    return True
def display():
  for row in board:
    print(row[0], "|", row[1], "|", row[2])


```

```python

while True:


    display()
    print("Player 1 (X)")

    row = int(input("Enter row: "))
    col = int(input("Enter column: "))

    board[row][col] = 'X'

    if check_win('X'):
        display()
        print("Player 1 won")
        break

    if check_full():
        display()
        print("Match Draw")
        break


    display()
    print("Player 2 (O)")

    row = int(input("Enter row: "))
    col = int(input("Enter column: "))

    board[row][col] = 'O'

    if check_win('O'):
        display()
        print("Player 2 won")
        break

    if check_full():
        display()
        print("Match Draw")
        break
```
