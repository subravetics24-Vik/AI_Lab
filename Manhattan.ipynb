```python
import heapq

class PuzzleNode:
    def __init__(self, state, parent=None, action=None, g=0, h=0):
        self.state = state
        self.parent = parent
        self.action = action
        self.g = g
        self.h = h
        self.f = g + h


    def __lt__(self, other):
        return self.f < other.f

def get_misplaced_tiles(state, goal_state):
    misplaced = 0
    for i in range(len(state)):
        if state[i] != 0 and state[i] != goal_state[i]:
            misplaced += 1
    return misplaced

def get_neighbors(state):
    neighbors = []
    blank_idx = state.index(0)
    row, col = blank_idx // 3, blank_idx % 3

    moves = [
        (-1, 0, 'Up'),
        (1, 0, 'Down'),
        (0, -1, 'Left'),
        (0, 1, 'Right')
    ]

    for r_offset, c_offset, action in moves:
        new_r, new_c = row + r_offset, col + c_offset
        if 0 <= new_r < 3 and 0 <= new_c < 3:
            new_blank_idx = new_r * 3 + new_c
            new_state = list(state)
            new_state[blank_idx], new_state[new_blank_idx] = new_state[new_blank_idx], new_state[blank_idx]
            neighbors.append((tuple(new_state), action))

    return neighbors

def solve_a_star(start_state, goal_state):
    """Solves the 8-puzzle using A* search."""
    start_h = get_misplaced_tiles(start_state, goal_state)
    start_node = PuzzleNode(start_state, g=0, h=start_h)

    open_list = []
    heapq.heappush(open_list, start_node)

    closed_set = set()

    nodes_expanded = 0

    while open_list:
        current_node = heapq.heappop(open_list)
        nodes_expanded += 1

        if current_node.state == goal_state:
            path = []
            curr = current_node
            while curr.parent is not None:
                path.append((curr.action, curr.state))
                curr = curr.parent
            path.reverse()
            return path, nodes_expanded

        closed_set.add(current_node.state)

        for next_state, action in get_neighbors(current_node.state):
            if next_state in closed_set:
                continue

            g_new = current_node.g + 1
            h_new = get_misplaced_tiles(next_state, goal_state)
            neighbor_node = PuzzleNode(next_state, parent=current_node, action=action, g=g_new, h=h_new)

            heapq.heappush(open_list, neighbor_node)

    return None, nodes_expanded

def print_board(state):
    for i in range(0, 9, 3):
        print(f"{state[i]} {state[i+1]} {state[i+2]}")
    print()
```

```python

start = (2, 8, 3,
         1, 6, 4,
         0, 7, 5)

goal = (1, 2, 3,
        8, 0, 4,
        7, 6, 5)

print("Start State:")
print_board(start)

print("Goal State:")
print_board(goal)

path, expanded = solve_a_star(start, goal)

if path:
    print(f"Solved in {len(path)} moves. Nodes expanded: {expanded}\n")
    for step, (action, state) in enumerate(path, 1):
        print(f"Step {step}: Move {action}")
        print_board(state)
else:
    print("No solution found.")
```
